```python
import numpy as np
# 1-D Array
arr1=np.arange(1,20)
arr1
```

```python
#creating array from the list
elem=[1,2,3,4,5]
print(np.array(elem))
print("With Step size=2 ",np.arange(1,10,2))
```

```python
#Arrays of ones and Zeros
print(np.ones(7)) #  creates an array filled with 1
print(np.zeros(7)) # creates an array filled with 0
print(np.full(7,4)) # creates an array of seven elements all filled with 4

```

```python
# Array using Random Numbers
print(np.random.rand(5)) #creates an array of 5 random values
print(np.random.randint(1,10,5)) # takes low, high and number of values
```

```python
                                                    Multidimensional  Array
```

```python
print(np.arange(12).reshape(4,3))
```

```python
print(np.array([[1,2,3,4],[2,3,4,5],[3,4,5,6],[4,5,6,7]])) # 2-D array
print(np.array([[[1,2],[2,4],[4,6]]]))
```

```python
print(np.ones(9).reshape(3,3))
print(np.zeros(16).reshape(4,4))
print(np.full(4,2).reshape(2,2))
```

```python
                                     Array Operations
```

```python
import numpy as np
a=np.arange(12).reshape(3,4)
b=np.random.randint(20,50,(3,4))
print("Addition",a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
print(a|b)
print(a&b)
print(a^b)
```

```python
print(f"Shape: {a.shape},   dim: {a.ndim},  size:{a.size}")
```

```python
#Reshaping the array
reshaped=b.reshape(4,3)
print(reshaped)
print(f"Transpose of reshaped:\n {reshaped.T}")
```

```python
# Slicing of array
print(b[2,2])
print(b[1:])
print(b[0,0:2])
```
